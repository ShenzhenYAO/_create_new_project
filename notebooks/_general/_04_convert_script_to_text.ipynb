This document is about adding .txt to conver files to .txt files as the Arena agent does not accept ipynb files, but .txt files

```python
import os
from pathlib import Path

def get_project_path():
    if "__file__" in globals():
        # Running as a .py script
        this_file_type = "python script"
        project_path = Path(__file__).resolve().parent.parent.parent.parent.parent
    else:
        # Presume running in a Jupyter notebook; use CWD’s great-grandparent
        this_file_type = "notebook"
        project_path = Path.cwd().resolve().parent.parent
    return project_path, this_file_type

project_path, this_file_type = get_project_path()
print("project_path:", project_path)
print("file tyep:", this_file_type)
```

```python
import os
import shutil

def copy_files(
    src_dir: str,
    dst_dir: str,
    src_prefix: str = "",
    src_ext: str = "",
    dst_suffix: str = "",
    dst_ext: str = None
) -> None:
    os.makedirs(dst_dir, exist_ok=True)
    
    if src_ext and not src_ext.startswith("."):
        src_ext = f".{src_ext}"
    if dst_ext and not dst_ext.startswith("."):
        dst_ext = f".{dst_ext}"

    for filename in os.listdir(src_dir):
        src_path = os.path.join(src_dir, filename)
        
        if not os.path.isfile(src_path):
            continue
        if src_prefix and not filename.startswith(src_prefix):
            continue
        if src_ext and not filename.endswith(src_ext):
            continue
            
        stem, current_ext = os.path.splitext(filename)
        final_ext = dst_ext if dst_ext is not None else current_ext
        new_filename = f"{stem}{dst_suffix}{final_ext}"
        dst_path = os.path.join(dst_dir, new_filename)
        
        shutil.copy2(src_path, dst_path)
```

```python
copy_files(
    src_dir=rf'{project_path}/backend/apps/python/41_job_application/modules',
    dst_dir=rf'{project_path}/backend/apps/python/41_job_application/modules/txt',
    src_prefix = "",
    src_ext = ".py",
    dst_suffix = ".py",
    dst_ext = "txt"
) 
```
